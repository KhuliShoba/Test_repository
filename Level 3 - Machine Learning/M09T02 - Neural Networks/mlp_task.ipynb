# Multi-Layer Perceptron (MLP) — Basketball Player Quality Evaluator

This notebook implements a simple MLP that takes **5 basketball performance ratings** as input and outputs a **quality score between 0 and 10**.

```python
# Cell 1 — Imports
import torch
import torch.nn as nn
```

```python
# Cell 2 — Sample input data
# Each row represents one player with 5 ratings:
# [shooting, dribbling, passing, defence, athleticism]  (all values 0–10)
sample_data = [
    [8.0, 7.5, 6.0, 5.5, 9.0],   # Player 1
    [5.0, 6.0, 8.5, 7.0, 6.5],   # Player 2
    [9.5, 8.0, 7.0, 4.0, 8.5],   # Player 3
    [4.5, 5.5, 6.5, 9.0, 7.0],   # Player 4
    [7.0, 7.0, 7.0, 7.0, 7.0],   # Player 5
]
```

```python
# Cell 3 — Convert to tensor
X_tensor = torch.tensor(sample_data, dtype=torch.float32)
print("Input tensor (X_tensor):")
print(X_tensor)
print("Shape:", X_tensor.shape)
```

```python
# Cell 4 — MLP class definition
class MLP(nn.Module):
    """Multi-Layer Perceptron for basketball player quality evaluation.

    Architecture:
        Input  layer : 5 neurons  (one per rating)
        Hidden layer : 8 neurons  + Sigmoid activation
        Hidden layer : 4 neurons  + Sigmoid activation
        Output layer : 1 neuron   + Sigmoid activation (scaled to 0-10)
    """

    def __init__(self):
        super(MLP, self).__init__()
        self.fc1 = nn.Linear(5, 8)
        self.fc2 = nn.Linear(8, 4)
        self.fc3 = nn.Linear(4, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        x = self.sigmoid(self.fc1(x))
        x = self.sigmoid(self.fc2(x))
        x = self.sigmoid(self.fc3(x))
        # Scale sigmoid output (0–1) to quality score range (0–10)
        return x * 10
```

```python
# Cell 5 — Architecture summary
# The network uses Sigmoid activations throughout so that each layer's output
# is bounded between 0 and 1 before the final scaling step.
# Input  (5) → Hidden1 (8, Sigmoid) → Hidden2 (4, Sigmoid) → Output (1, Sigmoid×10)
```

```python
# Cell 6 — Instantiate the MLP model
mlp = MLP()
print(mlp)
```

```python
# Cell 7 — Initialise weights using Xavier uniform initialisation
def init_weights(m):
    if isinstance(m, nn.Linear):
        nn.init.xavier_uniform_(m.weight)
        nn.init.zeros_(m.bias)

mlp.apply(init_weights)
print("Weights initialised successfully.")
for name, param in mlp.named_parameters():
    print(f"  {name}: {param.data}")
```

```python
# Cell 8 — Validate forward propagation
# Pass the sample input tensor through the network and print the predictions.
with torch.no_grad():
    output = mlp(X_tensor)

print("Output tensor (predicted quality scores):")
print(output)
```
